- Create a base class `Shape` with a method `area()` that raises `NotImplementedError`.
- Create subclasses `Rectangle` and `Circle` that implement `area()` appropriately.
- Instantiate both and print their areas.

````python
import math

class Shape:
    def area(self):
        raise NotImplementedError("Subclasses must implement abstract method")

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * (self.radius ** 2)

rect = Rectangle(4, 5)
circle = Circle(3)

print("Rectangle Area:", rect.area())
print("Circle Area:", round(circle.area(), 2))
````

- Create a base class `BankAccount` with `owner`, `balance`, and methods `deposit()` and `withdraw()`.
- Create a subclass `SavingsAccount` that adds an `interest_rate` and a method `add_interest()` which increases the balance.
- Demonstrate deposits, withdrawals, and interest application.

````python
class BankAccount:
    def __init__(self, owner, balance=0.0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited ${amount}. New Balance: ${self.balance}")

    def withdraw(self, amount):
        if 0 < amount <= self.balance:
            self.balance -= amount
            print(f"Withdrew ${amount}. Remaining Balance: ${self.balance}")
        else:
            print("Insufficient funds or invalid amount.")

class SavingsAccount(BankAccount):
    def __init__(self, owner, balance=0.0, interest_rate=0.05):
        super().__init__(owner, balance)
        self.interest_rate = interest_rate

    def add_interest(self):
        interest = self.balance * self.interest_rate
        self.balance += interest
        print(f"Added ${interest} interest. New Balance: ${self.balance}")

account = SavingsAccount("Alice", 1000, 0.05)
account.deposit(500)
account.withdraw(200)
account.add_interest()
````

- Create a class `CountDown` that takes a number `n` and counts down to 1.  
- Implement the iterator protocol (`__iter__()` and `__next__()` methods).  
- Use it in a `for` loop to print numbers in reverse order.

````python
class CountDown:
    def __init__(self, n):
        self.current = n

    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= 0:
            raise StopIteration
        val = self.current
        self.current -= 1
        return val

for num in CountDown(5):
    print(num)
````

- Create a custom iterator `FileLineReader` that takes a filename and reads one line at a time.  
- Implement `__iter__()` and `__next__()` so it can be used in a `for` loop.  
- Test it by reading from a sample `.txt` file and printing each line.

````python
class FileLineReader:
    def __init__(self, filename):
        self.filename = filename
        self.file = None

    def __iter__(self):
        self.file = open(self.filename, 'r')
        return self

    def __next__(self):
        line = self.file.readline()
        if not line:
            self.file.close()
            raise StopIteration
            
        return line.strip()

# Create a sample .txt file
sample_filename = "sample.txt"
with open(sample_filename, "w") as f:
    f.write("First line of the file.\n")
    f.write("Second line of the file.\n")
    f.write("Third line of the file.\n")

# Test FileLineReader in a for loop
reader = FileLineReader(sample_filename)

print("=== Reading File Lines ===")
for line in reader:
    print(line)
````

- Create an abstract class `Shape` with an abstract method `area()`.  
- Create subclasses `Rectangle` and `Circle` that implement the `area()` method.  
- Create a list of different shapes and loop through them, printing their area without checking their type.


````python
from abc import ABC, abstractmethod
import math

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Rectangle(Shape):
    def __init__(self, w, h):
        self.w = w
        self.h = h

    def area(self):
        return self.w * self.h

class Circle(Shape):
    def __init__(self, r):
        self.r = r

    def area(self):
        return math.pi * (self.r ** 2)

shapes = [Rectangle(10, 5), Circle(7), Rectangle(3, 4)]
for shape in shapes:
    print(f"Area: {shape.area():.2f}")
````

- a nonlocal variable modified inside a nested function.


````python
global_val = 100

def outer_scope():
    enclosing_val = "Outer"
    
    def inner_scope():
        nonlocal enclosing_val
        global global_val
        
        enclosing_val = "Modified Outer"
        global_val = 500

    inner_scope()
    print("Enclosing value after inner:", enclosing_val)

outer_scope()
print("Global value after inner:", global_val)
````

Create two classes: `Flyer` (method `move()` returns `"Flying"`) and `Swimmer` (method `move()` returns `"Swimming"`).  
Create a class `Duck` inheriting from both.  
- Use `super()` to decide which `move()` executes.  
- Explain the **Method Resolution Order (MRO)**.  

````python
class Flyer:
    def move(self):
        return "Flying"

class Swimmer:
    def move(self):
        return "Swimming"

class Duck(Flyer, Swimmer):
    def move(self):
        return super().move()

duck = Duck()
print("Duck move action:", duck.move())
print("Method Resolution Order:", [cls.__name__ for cls in Duck.__mro__])
````

Python resolves attributes and methods using the C3 Linearization algorithm. In class Duck(Flyer, Swimmer), Python searches in this order: Duck $\rightarrow$ Flyer $\rightarrow$ Swimmer $\rightarrow$ object. super().move() calls Flyer.move() because Flyer is listed first in the parent class declaration.

Create a class `Fibonacci` that generates Fibonacci numbers up to a given `n` using an iterator protocol.

````python
class Fibonacci:
    def __init__(self, n):
        self.n = n
        self.a, self.b = 0, 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.n:
            raise StopIteration
        val = self.a
        self.a, self.b = self.b, self.a + self.b
        self.count += 1
        return val

for fib in Fibonacci(8):
    print(fib, end=" ")
print()
````

# Library Management System  
  
## Scenario  
You are building a small library system to manage books, members, and borrowing operations. The system should support multiple types of books, track borrowed books, and allow iteration over members and books. The system should also demonstrate proper use of scopes and encapsulation.  
  
## Requirements and Concepts to Cover  
  
### 1. OOP & Inheritance  
- Create a base class `Book` with attributes: `title`, `author`, `year`.  
- Create subclasses:  
  - `EBook` with `file_size` and `format`.  
  - `PrintedBook` with `weight` and `dimensions`.  
- Create a base class `Member` with `name` and `member_id`.  
- Create subclasses:  
  - `StudentMember` with `grade_level`.  
  - `TeacherMember` with `department`.  
  
### 2. Polymorphism  
- Define a method `get_description()` in `Book` and override it in both `EBook` and `PrintedBook`.  
- Define a method `borrow_book()` in `Member` that behaves differently for `StudentMember` and `TeacherMember` (e.g., different borrowing limits).  
  
### 3. Iterators  
- Implement an iterator for the `Library` class to iterate over all books.  
- Implement an iterator for the `Member` class to iterate over borrowed books.  
  
### 4. Scope  
- Use private and protected attributes to demonstrate scope:  
  - `_borrowed_books` in `Member` (protected).  
  - `__library_name` in `Library` (private).  
- Provide getter and setter methods where appropriate to access private/protected attributes.  
  
### 5. Extra Features  
- Track which books are available and which are borrowed.  
- Allow searching books by title or author.  
- Count the total number of books borrowed by all members (use a class-level variable).  

````python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year
        self.is_borrowed = False

    def get_description(self):
        return f"'{self.title}' by {self.author} ({self.year})"

class EBook(Book):
    def __init__(self, title, author, year, file_size, format_type):
        super().__init__(title, author, year)
        self.file_size = file_size
        self.format_type = format_type

    def get_description(self):
        return f"EBook: '{self.title}' by {self.author} [{self.file_size}MB, {self.format_type}]"

class PrintedBook(Book):
    def __init__(self, title, author, year, weight, dimensions):
        super().__init__(title, author, year)
        self.weight = weight
        self.dimensions = dimensions

    def get_description(self):
        return f"Printed Book: '{self.title}' by {self.author} [{self.weight}kg, {self.dimensions}]"

class Member:
    total_borrowed_books_count = 0  

    def __init__(self, name, member_id):
        self.name = name
        self.member_id = member_id
        self._borrowed_books = []  

    def borrow_book(self, book):
        if not book.is_borrowed:
            book.is_borrowed = True
            self._borrowed_books.append(book)
            Member.total_borrowed_books_count += 1
            print(f"{self.name} borrowed '{book.title}'")
            return True
        print(f"'{book.title}' is currently unavailable.")
        return False

    def get_borrowed_books(self):
        return [b.get_description() for b in self._borrowed_books]

    def __iter__(self):
        return iter(self._borrowed_books)

class StudentMember(Member):
    def __init__(self, name, member_id, grade_level, max_limit=2):
        super().__init__(name, member_id)
        self.grade_level = grade_level
        self.max_limit = max_limit

    def borrow_book(self, book):
        if len(self._borrowed_books) >= self.max_limit:
            print(f"Borrowing limit reached for Student {self.name} (Max: {self.max_limit}).")
            return False
        return super().borrow_book(book)

class TeacherMember(Member):
    def __init__(self, name, member_id, department, max_limit=5):
        super().__init__(name, member_id)
        self.department = department
        self.max_limit = max_limit

    def borrow_book(self, book):
        if len(self._borrowed_books) >= self.max_limit:
            print(f"Borrowing limit reached for Teacher {self.name} (Max: {self.max_limit}).")
            return False
        return super().borrow_book(book)

class Library:
    def __init__(self, name):
        self.__library_name = name  
        self.books = []

    def get_library_name(self):
        return self.__library_name

    def add_book(self, book):
        self.books.append(book)

    def search_books(self, query):
        query = query.lower()
        return [b for b in self.books if query in b.title.lower() or query in b.author.lower()]

    def __iter__(self):
        return iter(self.books)


lib = Library("City Central Library")

b1 = EBook("Python 101", "Guido van Rossum", 2020, 15, "PDF")
b2 = PrintedBook("Clean Code", "Robert C. Martin", 2008, 1.2, "15x23 cm")
b3 = EBook("Data Structures", "Jane Doe", 2021, 5, "EPUB")

lib.add_book(b1)
lib.add_book(b2)
lib.add_book(b3)

student = StudentMember("Alex", "S101", "10th Grade")
teacher = TeacherMember("Dr. Smith", "T501", "Computer Science")

print(f"=== {lib.get_library_name()} Setup ===")
student.borrow_book(b1)
student.borrow_book(b2)
student.borrow_book(b3)  

teacher.borrow_book(b3)

print("\n--- Books Borrowed by Alex ---")
for book in student:
    print(book.get_description())

print("\n--- Total System Borrowed Book Count ---")
print("Total Borrowed across members:", Member.total_borrowed_books_count)
````
