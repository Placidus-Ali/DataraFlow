```python
# 1. Write a program that stores a list of student records (name, age, grade) 
# into a JSON file, and then reads the file back into a Python object.

import json

# Sample student records
students = [
    {"name": "Alice", "age": 20, "grade": "A"},
    {"name": "Bob", "age": 22, "grade": "B"},
    {"name": "Charlie", "age": 21, "grade": "A"}
]

filename = "students.json"

# Write records to JSON file
with open(filename, "w") as file:
    json.dump(students, file, indent=4)
print("Data successfully written to file.")

# Read records back into Python
with open(filename, "r") as file:
    loaded_students = json.load(file)

print("\nData read back from file:")
print(loaded_students)
```

```python
# 2. Build a program that asks the user 5 random math questions 
# (addition, subtraction, multiplication, division) and keeps score of correct answers.

import random

score = 0
operators = ['+', '-', '*']

print("=== Quick Math Quiz ===")

for i in range(1, 6):
    num1 = random.randint(1, 10)
    num2 = random.randint(1, 10)
    op = random.choice(operators)
    
    correct_answer = eval(f"{num1} {op} {num2}")
    user_answer = input(f"Question {i}: What is {num1} {op} {num2}? ")
    
    if float(user_answer) == float(correct_answer):
        print("Correct!\n")
        score += 1
    else:
        print(f"Incorrect. The correct answer was {correct_answer}\n")

print(f"Quiz Finished! Your Score: {score}/5")
```

```python
# 3. Create a program that asks the user for a number and divides 100 by that number. 
# Use try/except to handle invalid input and division by zero.

user_input = input("Enter a number to divide 100 by: ")

try:
    number = float(user_input)
    result = 100 / number
    print(f"100 divided by {number} is: {result}")
except ZeroDivisionError:
    print("Error: Cannot divide 100 by zero.")
except ValueError:
    print("Error: Invalid input! Please enter a valid number (e.g., 5 or 2.5).")
```

```python
#  4. Write a program that takes a user’s birthday as input and tells them 
# how many days are left until their next birthday.

import datetime

bday_input = input("Enter your birthday (YYYY-MM-DD): ")

try:
    birthday = datetime.datetime.strptime(bday_input, "%Y-%m-%d").date()
    today = datetime.date.today()
    
    # Set birthday for current year
    next_birthday = datetime.date(today.year, birthday.month, birthday.day)
    
    # If birthday passed this year, calculate for next year
    if next_birthday < today:
        next_birthday = datetime.date(today.year + 1, birthday.month, birthday.day)
        
    days_left = (next_birthday - today).days
    print(f"There are {days_left} days left until your next birthday!")

except ValueError:
    print("Invalid date format. Please use YYYY-MM-DD (e.g., 2002-08-15).")
```

1. What is the difference between a Python module and a package?


- Module: A single Python file ending in .py that contains reusable code (functions, classes, variables). 
  Example: math.py or datetime.py

- Package: A directory (folder) that contains multiple Python modules grouped together, 
  along with an __init__.py file that tells Python to treat the directory as a package.

8. Explain the difference between import module and from module import function.


1. import module:
   Imports the entire module. To use any function from it, you must prefix it with the module name.
   Example:
     import math
     print(math.sqrt(16))

2. from module import function:
   Imports only a specific function directly into your script. You can use it directly without prefixing.
   Example:
     from math import sqrt
     print(sqrt(16))

10. What is a virtual environment in Python, and why is it useful?

A virtual environment is an isolated, self-contained workspace directory for a specific Python project.

Why it is useful:
1. Prevents Dependency Conflicts: Different projects can use different versions of the same package 
   (e.g., Project A uses Django 3.0, Project B uses Django 4.0).
2. Keeps Global Python Clean: Keeps system-wide Python installation free from clutter.
3. Easy Reproducibility: Makes it simple to share requirements with team members (using requirements.txt).

12. Explain what happens when you try to open a JSON file with invalid JSON content.

import json

When you attempt to parse a file or string containing invalid JSON format (e.g., missing quotes, 
trailing commas, or incorrect brackets) using json.load() or json.loads(), 
Python raises a json.decoder.JSONDecodeError exception.


```python
# Example
invalid_json_str = '{"name": "Alice", "age": 20,}'  # Trailing comma makes it invalid

try:
    data = json.loads(invalid_json_str)
except json.decoder.JSONDecodeError as e:
    print("Error caught: Failed to decode invalid JSON format.")
    print("Details:", e)
```
